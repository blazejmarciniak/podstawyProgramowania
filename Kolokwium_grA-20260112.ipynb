#### Kryteria oceniania
Ocenie podlegają następujące elementy:
* poprawność proponowanego rozwiązania
* kompletność wykonania zadania
* czytelność kodu
* odpowiednie oznaczanie typów
* zabezpieczenie kodu przed niepoprawnym użyciem
* łatwość obsługi programu.

### Zadanie 9. (5 pkt.)

* fahrenheit - konwertuje aktualną temperaturę na stopnie Fahrenheita i zwraca ją do programu, metoda obsługiwana jak atrybut. Po przypisaniu nowej wartości konwertuje ją na stopnie Celsjusza i aktualizuje wartość aktualnej temperatury.

$$T_F = T_C * \frac{9}{5} + 32$$
